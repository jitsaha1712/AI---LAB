```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf
import cv2
from sklearn.preprocessing import LabelEncoder
print("="*60)
print("introduction to python libarys for ai")

```

```python
print("\n---NumPy---\n")
numbers = np.array([10,20,30,40,50])

print("array:",numbers)
print("sum",np.sum(numbers))
print("mean",np.mean(numbers))
print("max:",np.max(numbers))
print("min:",np.min(numbers))
```

```python
print("\n---Pandas---")

student = {
    "name":["Alice","bob","charle"],
    "marks":[80,65,90]
}
df = pd.DataFrame(student)
print("Student data\n")
print(df)
```

```python
print("\n---Matplotlib---")
print("Displaying Bar Chart...")

plt.figure(figsize=(6,4))
plt.bar(df["name"],df["marks"])
plt.title("student data")
plt.xlabel("name")
plt.ylabel("marks")
plt.show()
```

```python
print("\n---scikit-learn---")

encoder = LabelEncoder();

df["studentID"] = encoder.fit_transform(df["name"])

print("after encoding names into numbers")
print(df)
```

```python
print("\n...Tensor flow")

tensor1 = tf.constant([1,2,3])
tensor2 = tf.constant([4,5,6])

print("Tensor1:",tensor1.numpy())
print("Tensor2:",tensor2.numpy())

result = tf.add(tensor1,tensor2)
print("tensor addition:",result.numpy())
```

```python
print("\n---OpenCV---")
#CREATE A BLANK BLACK IMAGE
image = np.zeros((250,250,3),dtype=np.uint8)

#DRAW A BLUE RECTANGLE
cv2.rectangle(image,(50,50),(200,200),(255,0,0),3)

#DRAW A GREEN CIRCLE
cv2.circle(image,(125,125),40,(0,255,0),3)

#DISPLAY IMAGE
plt.imshow(cv2.cvtColor(image,cv2.COLOR_BGR2RGB))
plt.title("OpenCV Shapes")
plt.axis("off")
plt.show()
```
